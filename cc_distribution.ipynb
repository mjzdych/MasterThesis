```python

import os
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from matplotlib.patches import Patch
import geopandas as gpd
import xarray as xr
from scipy.ndimage import label as nd_label
```

```python
DATA_FILE = "full_processed_training_dataset.nc"
```

```python
print("Loading climatological CC from full dataset...")
ds = xr.open_dataset(DATA_FILE)
print(ds)
```

```python
print(list(ds.data_vars))
```

```python
cc_target = ds["CC_target_next_day"].values.ravel()
cc_target = cc_target[~np.isnan(cc_target)]

print("min:", cc_target.min())
print("max:", cc_target.max())
print("mean:", cc_target.mean())
print("unique value count:", len(np.unique(cc_target)))
print("fraction == 0:", (cc_target == 0).mean())
print("fraction == 1:", (cc_target == 1).mean())
for q in [0.25, 0.40, 0.50, 0.70, 0.90, 0.99]:
    print(f"  q{int(q*100)}: {np.quantile(cc_target, q):.6f}")
ds.close()
```
