## Feature Ablation comparison

```python
from scipy.stats import friedmanchisquare, wilcoxon
import numpy as np

cn_only   = [0.567,0.650,0.513,0.668,0.517,0.564,0.638,0.562,0.578,0.382]
era5_only = [-0.150,0.174,0.018,-0.020,-0.011,0.021,-0.293,-0.105,-0.162,-0.011]
cn_era5   = [0.564,0.650,0.520,0.676,0.515,0.575,0.635,0.562,0.594,0.409]

stat, p = friedmanchisquare(cn_only, era5_only, cn_era5)
print(f"Friedman: stat={stat:.3f}, p={p:.6f}")

for a, b, name in [(cn_era5,era5_only,"CN+ERA5 vs ERA5"),
                   (cn_only,era5_only,"CN-only vs ERA5"),
                   (cn_era5,cn_only,"CN+ERA5 vs CN-only")]:
    s,pv = wilcoxon(a, b)
    print(f"{name}: p={min(pv*3,1):.6f}")
```
